````python
# Load and explore the banking dataset
data = pd.read_csv("../../data/banking.csv")

print(f"Dataset dimensions: {len(data)} rows and {data.shape[1]} columns\n")

# Display structure
print(data.info())

# Summary statistics
print("\n", data.describe())

# Check for missing values
print("\nMissing values per column:")
print(data.isnull().sum())

# Look at the target variable distribution
print("\nTarget variable (y) distribution:")
print(data['y'].value_counts())
print("\n", data['y'].value_counts(normalize=True))
````

## Data Preparation

Before building models, we need to prepare our data properly.


````python
# Convert categorical variables to appropriate types
# Note: The dataset has numeric y (0/1) and is missing some categorical columns
categorical_vars = ["marital", "education", "housing", "loan", "contact", "poutcome"]
for col in categorical_vars:
    if col in data.columns:
        data[col] = data[col].astype('category')

# Convert y to categorical for stratification
data['y'] = data['y'].astype('category')

# Create train/test split (80/20)
np.random.seed(123)  # For reproducibility
from sklearn.model_selection import train_test_split
train_data, test_data = train_test_split(data, test_size=0.2, random_state=123, stratify=data['y'])

print(f"Training set: {len(train_data)} observations")
print(f"Test set: {len(test_data)} observations")

# Check that target variable is balanced in both sets
print("\nTarget distribution in training set:")
print(train_data['y'].value_counts(normalize=True))
print("\nTarget distribution in test set:")
print(test_data['y'].value_counts(normalize=True))
````

## 1. Lasso Regression

**What is Lasso?** Lasso (Least Absolute Shrinkage and Selection Operator) is a linear model that automatically selects the most important features by setting less important coefficients to zero. This helps prevent overfitting and makes the model easier to interpret.

**When to use Lasso:**
- When you have many features and want automatic feature selection
- When you need an interpretable model
- When you suspect many features are irrelevant

**Key implementation details:**
- We use `StandardScaler` to normalize features (important for L1 regularization)
- We use the `liblinear` solver which is optimized for smaller datasets and converges reliably
- The regularization parameter `C` controls the strength of regularization (smaller C = stronger regularization)

### Logistic Lasso: Predicting customer subscription (y)

````python
# Prepare data for logistic lasso (training set)
from sklearn.preprocessing import StandardScaler

# Prepare X (features) and y (target)
X_train = pd.get_dummies(train_data.drop('y', axis=1), drop_first=True)
y_train = train_data['y'].astype(int)

X_test = pd.get_dummies(test_data.drop('y', axis=1), drop_first=True)
y_test = test_data['y'].astype(int)

# Align columns between train and test
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)

# Scale features (important for L1 regularization convergence)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Fit logistic lasso using a simpler, more stable approach
# Use liblinear solver which is better for small datasets
from sklearn.linear_model import LogisticRegression

print("Training Lasso (L1 regularized logistic regression)...")
lasso_model = LogisticRegression(
    penalty='l1',
    solver='liblinear',  # More stable than saga for this problem
    C=1.0,  # Inverse of regularization strength
    random_state=123,
    max_iter=1000
)
lasso_model.fit(X_train_scaled, y_train)

print(f"Regularization parameter (C): {lasso_model.C}")

# Get coefficients for the model
lasso_coef = pd.Series(lasso_model.coef_[0], index=X_train.columns)
selected_features = lasso_coef[lasso_coef != 0]
print(f"\nNumber of selected features: {len(selected_features)} out of {len(lasso_coef)}")

# Show top features by absolute coefficient value
print("\nTop 10 most important features (by |coefficient|):")
print(lasso_coef.abs().sort_values(ascending=False).head(10))

# Make predictions on test set
lasso_pred_prob = lasso_model.predict_proba(X_test_scaled)[:, 1]
lasso_pred_class = (lasso_pred_prob > 0.5).astype(int)

# Evaluate performance
lasso_accuracy = (lasso_pred_class == y_test).mean()
print(f"\nLasso Test Accuracy: {lasso_accuracy:.3f}")

# Confusion matrix
from sklearn.metrics import confusion_matrix
lasso_cm = confusion_matrix(y_test, lasso_pred_class)
print("\nConfusion Matrix:")
print(pd.DataFrame(lasso_cm, columns=['Predicted 0', 'Predicted 1'], 
                   index=['Actual 0', 'Actual 1']))
````

## 2. Decision Trees

**What are Decision Trees?** Decision trees make predictions by asking a series of yes/no questions about the features. They're like a flowchart that leads to a prediction.

**When to use Decision Trees:**
- When you need a highly interpretable model
- When relationships between features are non-linear
- When you want to understand the decision-making process

**Pros:** Easy to interpret, handles non-linear relationships
**Cons:** Can overfit, unstable (small data changes can create very different trees)


````python
# Fit a decision tree for classification
from sklearn.tree import DecisionTreeClassifier, plot_tree

tree_mod = DecisionTreeClassifier(min_samples_split=20, ccp_alpha=0.01, random_state=123)
tree_mod.fit(X_train, y_train)

# Plot the tree
plt.figure(figsize=(20, 10))
plot_tree(tree_mod, feature_names=X_train.columns, class_names=['0', '1'], 
          filled=True, rounded=True, fontsize=8)
plt.title("Decision Tree for Customer Subscription Prediction")
plt.show()

# Make predictions
tree_pred_class = tree_mod.predict(X_test)
tree_pred_prob = tree_mod.predict_proba(X_test)[:, 1]

# Evaluate performance
tree_accuracy = (tree_pred_class == y_test).mean()
print(f"\nDecision Tree Test Accuracy: {tree_accuracy:.3f}")

# Confusion matrix
tree_cm = confusion_matrix(y_test, tree_pred_class)
print("\nConfusion Matrix:")
print(pd.DataFrame(tree_cm, columns=['Predicted 0', 'Predicted 1'], 
                   index=['Actual 0', 'Actual 1']))

# Feature importance
print("\nFeature Importance (Decision Tree) - Top 10:")
importance_tree = pd.Series(tree_mod.feature_importances_, index=X_train.columns)
print(importance_tree.sort_values(ascending=False).head(10))
````

## 3. Random Forests

**What are Random Forests?** Random Forests combine many decision trees, where each tree is trained on a random subset of the data and features. The final prediction is the average (or majority vote) of all trees.

**When to use Random Forests:**
- When you want better accuracy than a single decision tree
- When you have enough data (works well with large datasets)
- When you want feature importance rankings

**Pros:** Usually more accurate than single trees, provides feature importance, handles missing values
**Cons:** Less interpretable than single trees, can overfit with very noisy data


````python
# Fit a random forest for classification
from sklearn.ensemble import RandomForestClassifier

np.random.seed(123)
mtry = int(np.sqrt(X_train.shape[1]))  # Standard mtry for classification
rf_mod = RandomForestClassifier(n_estimators=500, max_features=mtry, random_state=123, n_jobs=-1)
rf_mod.fit(X_train, y_train)

print(f"Random Forest with {rf_mod.n_estimators} trees")
print(f"Out-of-bag score: {rf_mod.oob_score_:.3f}" if hasattr(rf_mod, 'oob_score_') else "")

# Make predictions
rf_pred_class = rf_mod.predict(X_test)
rf_pred_prob = rf_mod.predict_proba(X_test)[:, 1]

# Evaluate performance
rf_accuracy = (rf_pred_class == y_test).mean()
print(f"\nRandom Forest Test Accuracy: {rf_accuracy:.3f}")

# Confusion matrix
rf_cm = confusion_matrix(y_test, rf_pred_class)
print("\nConfusion Matrix:")
print(pd.DataFrame(rf_cm, columns=['Predicted No', 'Predicted Yes'], 
                   index=['Actual No', 'Actual Yes']))

# Feature importance (top 10)
importance_rf = pd.Series(rf_mod.feature_importances_, index=X_train.columns)
print("\nTop 10 Most Important Features (Random Forest):")
print(importance_rf.sort_values(ascending=False).head(10))

# Plot variable importance
plt.figure(figsize=(10, 6))
importance_rf.sort_values(ascending=False).head(15).plot(kind='barh')
plt.xlabel('Feature Importance')
plt.title('Random Forest Feature Importance (Top 15)')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
````

## 4. Gradient Boosting (using scikit-learn)

**What is Gradient Boosting?** Gradient boosting builds models sequentially, where each new model tries to correct the mistakes of the previous models. Scikit-learn's `GradientBoostingClassifier` provides a robust implementation with feature importance support.

**When to use Gradient Boosting:**
- When you want state-of-the-art predictive performance
- When you have sufficient data and computational resources
- When accuracy is more important than interpretability
- When you need feature importance rankings

**Pros:** Often achieves the best predictive performance, handles missing values, provides feature importance, built into scikit-learn (no external dependencies)
**Cons:** More complex to tune, less interpretable, slower to train than simpler models

````python
# Use scikit-learn's Gradient Boosting (no external dependencies needed)
from sklearn.ensemble import GradientBoostingClassifier

# Fit gradient boosting model
print("Training Gradient Boosting model...")
gb_mod = GradientBoostingClassifier(
    n_estimators=100,
    max_depth=4,
    learning_rate=0.1,
    subsample=0.8,
    random_state=123
)

gb_mod.fit(X_train, y_train)

print(f"Number of boosting stages: {gb_mod.n_estimators_}")

# Feature importance
importance_df = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': gb_mod.feature_importances_
})
importance_df = importance_df.sort_values('Importance', ascending=False)

# Plot feature importance
plt.figure(figsize=(10, 6))
importance_df.head(15).plot(x='Feature', y='Importance', kind='barh', legend=False)
plt.xlabel('Importance')
plt.title('Gradient Boosting Feature Importance (Top 15)')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

# Make predictions
gb_pred_prob = gb_mod.predict_proba(X_test)[:, 1]
gb_pred_class = (gb_pred_prob > 0.5).astype(int)

# Evaluate performance
gb_accuracy = (gb_pred_class == y_test).mean()
print(f"\nGradient Boosting Test Accuracy: {gb_accuracy:.3f}")

# Confusion matrix
gb_cm = confusion_matrix(y_test, gb_pred_class)
print("\nConfusion Matrix:")
print(pd.DataFrame(gb_cm, columns=['Predicted 0', 'Predicted 1'], 
                   index=['Actual 0', 'Actual 1']))

# ROC curve and AUC
from sklearn.metrics import roc_curve, auc
fpr, tpr, thresholds = roc_curve(y_test, gb_pred_prob)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.3f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve (Gradient Boosting)')
plt.legend(loc="lower right")
plt.show()

print(f"\nAUC: {roc_auc:.3f}")
````

## Why Causal Inference Matters in Finance

**The Problem with Predictive Models:** Traditional ML models excel at prediction but struggle with **causation**. In finance, we often need to answer questions like:

- "What would happen if we **change** our interest rates?"
- "What is the **causal effect** of a marketing campaign on customer acquisition?"
- "How much **additional revenue** would we generate from a new product feature?"

**Correlation ≠ Causation:** Predictive models find patterns but can't tell us what happens when we intervene. Causal ML bridges this gap.

## Key Concepts in Causal Machine Learning

### 1. **Treatment Effects**
- **Average Treatment Effect (ATE)**: Average impact of treatment across population
- **Conditional Average Treatment Effect (CATE)**: Treatment effect for specific subgroups
- **Individual Treatment Effect (ITE)**: Personalized treatment effects

### 2. **Confounding**
Variables that affect both treatment assignment and outcome, creating spurious correlations.

### 3. **Identification Strategies**
- **Randomized Experiments**: Gold standard but often impractical
- **Natural Experiments**: Exploit random variation in real-world settings
- **Instrumental Variables**: Use variables that affect treatment but not outcome directly
- **Regression Discontinuity**: Exploit arbitrary cutoffs in treatment assignment

## Modern Causal ML Methods

### 1. **Double/Debiased Machine Learning (DML)**
Combines ML flexibility with causal inference rigor by:
- Using ML to model nuisance parameters
- Applying cross-fitting to avoid overfitting bias
- Providing valid confidence intervals

### 2. **Causal Forests**
Extension of Random Forests for heterogeneous treatment effects:
- Estimates personalized treatment effects
- Provides uncertainty quantification
- Handles high-dimensional data

### 3. **Meta-Learners**
- **T-Learner**: Separate models for treated/control groups
- **S-Learner**: Single model with treatment as feature
- **X-Learner**: Combines T-learner predictions optimally
- **R-Learner**: Uses residual-based approach

## Financial Applications

### 1. **Credit Risk & Lending**
- **Question**: "What's the causal effect of credit limit increases on default risk?"
- **Challenge**: Customers who get increases may be systematically different
- **Solution**: Use causal ML to control for selection bias

### 2. **Marketing & Customer Acquisition**
- **Question**: "Which customers benefit most from promotional offers?"
- **Application**: Personalized treatment effects for targeted campaigns
- **Business Value**: Optimize marketing spend and ROI

### 3. **Product Pricing**
- **Question**: "How do price changes affect demand across customer segments?"
- **Application**: Heterogeneous price elasticity estimation
- **Business Value**: Dynamic pricing strategies

### 4. **Regulatory Compliance**
- **Question**: "Do our algorithms create disparate impact across protected groups?"
- **Application**: Causal fairness analysis
- **Business Value**: Ensure fair lending practices

### Method 3: Meta-Learners (T-Learner and X-Learner)

Meta-learners use different strategies to combine ML models for causal inference.


**Optional: T-Learner Meta-Learner**

*To run this cell, convert it to Code.*

```python
# T-Learner: Separate models for treated and control
from sklearn.ensemble import RandomForestRegressor

treated_data = causal_data_numeric[causal_data_numeric['treatment'] == 1].drop(['treatment'], axis=1)
control_data = causal_data_numeric[causal_data_numeric['treatment'] == 0].drop(['treatment'], axis=1)

# Fit separate models
model_treated = RandomForestRegressor(n_estimators=500, random_state=123)
model_control = RandomForestRegressor(n_estimators=500, random_state=123)

model_treated.fit(treated_data.drop('outcome', axis=1), treated_data['outcome'])
model_control.fit(control_data.drop('outcome', axis=1), control_data['outcome'])

# Predict for all observations
X_all = causal_data_numeric.drop(['treatment', 'outcome'], axis=1)
mu1_hat = model_treated.predict(X_all)  # Potential outcome if treated
mu0_hat = model_control.predict(X_all)  # Potential outcome if control

# Individual treatment effects
tau_tlearner = mu1_hat - mu0_hat
ate_tlearner = tau_tlearner.mean()

print(f"T-Learner ATE: {ate_tlearner:.3f}")
print(f"T-Learner ITE std dev: {tau_tlearner.std():.3f}")

# Compare methods
comparison_df = pd.DataFrame({
    'Method': ['Naive', 'Causal Forest', 'Double ML', 'T-Learner'],
    'ATE': [naive_ate, ate_cf, ate_dml, ate_tlearner]
})
comparison_df['ATE'] = comparison_df['ATE'].round(3)

print("\nMethod Comparison:")
print(comparison_df.to_string(index=False))

# Plot comparison
plt.figure(figsize=(10, 6))
colors = ['red', 'green', 'blue', 'orange']
plt.barh(comparison_df['Method'], comparison_df['ATE'], color=colors)
plt.xlabel('ATE')
plt.ylabel('Method')
plt.title('Average Treatment Effect Estimates')
plt.axvline(x=0, color='black', linestyle='--', linewidth=1)
for i, (method, ate) in enumerate(zip(comparison_df['Method'], comparison_df['ATE'])):
    plt.text(ate + 0.005 if ate >= 0 else ate - 0.005, i, f'{ate:.3f}', 
             va='center', ha='left' if ate >= 0 else 'right')
plt.tight_layout()
plt.show()
```
